### __Description: This is an example of how to use the simulator of the QIP System__

<font color=yellow>(Notice) Start the simulation first! Use below command __in another terminal__:</font>  

./simulator/base --ctrl_ms [user specified control period (ms)]  

ex) ./simulator/base --ctrl_ms 10

```python
from fill_the_code import Controller
import socket

def send_command(command):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        s.connect(('localhost', 65432))
        s.sendall(command.encode())
        data = s.recv(1024).decode()
    return data.strip()
```

#### Participants may follow the steps.
1. <font color=skyblue>_your_controller_</font> : Instantiate your controller in the 'fill_the_code.py' file. (see the documentation for details.)
2. <font color=skyblue>_CTRL_MS_</font> : Specify the control period, which is defined in milliseconds. <font color=yellow>_Set the same value as the simulation._ </font>
3. <font color=skyblue>_MAX_STEPS_</font> : Specify the time steps. For example, the one below is set to 10 seconds.

```python
your_controller = Controller()                               # Instantiate your controller in fill_the_code.py.
CTRL_MS = 10                                                 # Control period in milliseconds.
MAX_STEPS = int(10 * 1000 / CTRL_MS)                         # Maximum number of steps (e.g., 10 sec total).
```

#### Simulate the QIP System.

```python
steps = 0  
state_str = send_command('reset')                                                          # Initialize the QIP system and get the initial state.
state = list(map(float, state_str.split()))                                                # Make the state a list.

while True:
    steps += 1
    force = Controller().compute_force(state)                                              # Compute the proper force from the controller.

    next_state_str = send_command(f'step {force}')                                         # Get the next state (changed from the applied force).
    next_state = list(map(float, next_state_str.split()))                                  # Make the next state a list.
    print(f"Steps:{steps}, State: {state}, Force: {force}, Next State: {next_state}")      # Print the state, action, and next state.
   
    state = next_state                                                                     # Update the state.
    if steps >= MAX_STEPS:                                                                 # break the loop.
        break
```
